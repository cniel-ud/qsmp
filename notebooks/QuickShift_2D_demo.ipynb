```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({
    "text.usetex": True,    
    "font.size": 12
})
img_dir = Path('../results/img/')
```

```python
delta = 0.025
mu_1 = [-1, -1]
mu_2 = [1, 1]
sigma_1 = [2, 1]
sigma_2 = [0.5, 0.3]
rho_1 = 0.5
rho_2 = 0.5
x = np.arange(-3, 3, delta)
y = np.arange(-3, 3, delta)
X, Y = np.meshgrid(x, y)
z_1 = (X - mu_1[0])**2/sigma_1[0]**2 + (Y - mu_1[1])**2/sigma_1[1]**2 - 2*rho_1*(X-mu_1[0])*(Y-mu_1[1])/np.prod(sigma_1)
Z_1 = np.exp(-z_1/(2*(1-rho_1**2))) / (2*np.pi*np.prod(sigma_1)*np.sqrt(1-rho_1**2))
z_2 = (X - mu_2[0])**2/sigma_2[0]**2 + (Y - mu_2[1])**2/sigma_2[1]**2 - 2*rho_2*(X-mu_2[0])*(Y-mu_2[1])/np.prod(sigma_2)
Z_2 = np.exp(-z_2/(2*(1-rho_2**2))) / (2*np.pi*np.prod(sigma_2)*np.sqrt(1-rho_2**2))
Z = 20*Z_1 + Z_2
data = np.vstack([
    [-2, -1.5],
    [-2, -2],
    [-2.5, -1.1],
    [-1.5, -1.5],
    [1, 0.5],
    [0.7, 0.8],
    [1.5, 1],
    [0.7, 0.2],
    [0.2, -0.3],
    [-0.3, -1.1]
]
)
fig, ax = plt.subplots()
CS = ax.contour(X, Y, Z)
ax.scatter(data[:,0], data[:,1], c='r', zorder=3)
ax.axis('off')
fpath = img_dir.joinpath('QuickShift_cartoon.svg')
plt.savefig(fpath, bbox_inches='tight', pad_inches=0)
```
